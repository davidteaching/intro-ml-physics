## 1. NumPy: Basic Tensor Operations

```python
import numpy as np

# Create vectors and matrices
x = np.array([1.0, 2.0])
A = np.array([[1, 2], [3, 4]])

# Operations
y = x @ A          # matrix-vector product
u, s, vt = np.linalg.svd(A)  # SVD
eigvals, eigvecs = np.linalg.eig(A @ A.T)

print("x:", x)
print("A:\n", A)
print("x @ A:", y)
print("SVD singular values:", s)
print("Eigenvalues of A^T A:", eigvals)
```

# 2. scikit-learn: Standard ML Algorithms

```python
from sklearn.linear_model import LinearRegression
from sklearn.decomposition import PCA
from sklearn.cluster import KMeans
from sklearn.gaussian_process import GaussianProcessRegressor

# Generate synthetic data
np.random.seed(42)
X = np.random.randn(100, 5)
y = X[:, 0] + 0.5 * X[:, 1] + 0.1 * np.random.randn(100)

# Fit a linear model
model = LinearRegression()
model.fit(X, y)
y_pred = model.predict(X)
print("Linear regression coefficients:", model.coef_)

# PCA
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X)
print("PCA explained variance ratio:", pca.explained_variance_ratio_)

# K-means
kmeans = KMeans(n_clusters=3, random_state=42, n_init=10)
labels = kmeans.fit_predict(X)
print("K-means cluster counts:", np.bincount(labels))
```

# 3. PyTorch: Deep Learning and Automatic Differentiation

```python
import torch
import torch.nn as nn
import torch.optim as optim

# Tensors and gradients
x = torch.tensor([1.0, 2.0], requires_grad=True)
y = (x ** 2).sum()
y.backward()
print("Gradients:", x.grad)

# Define a simple network
model = nn.Sequential(
    nn.Linear(10, 20),
    nn.ReLU(),
    nn.Linear(20, 1)
)
optimizer = optim.Adam(model.parameters(), lr=1e-3)

print("PyTorch model:")
print(model)
```

# 4. Qiskit: Quantum Circuits (Conceptual)

```python
try:
    from qiskit import QuantumCircuit
    from qiskit_aer import AerSimulator
    HAS_QISKIT = True
except ImportError:
    HAS_QISKIT = False

if HAS_QISKIT:
    # Create a Bell state
    qc = QuantumCircuit(2)
    qc.h(0)
    qc.cx(0, 1)
    qc.measure_all()
    
    sim = AerSimulator()
    job = sim.run(qc, shots=1024)
    counts = job.result().get_counts()
    print("Bell state measurement counts:", counts)
else:
    print("Qiskit not installed. Install with: pip install qiskit qiskit-aer")
    print("For a Bell state, the output would be approximately {'00': 512, '11': 512}")
```
